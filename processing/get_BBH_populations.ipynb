# Get BBH populations

This script extracts the BBH populations and reweights all of them using the model_weights to the same "final" population.



## Accretion_populations

```python
%load_ext autoreload
%autoreload 2
import numpy as np
from pathlib import Path
import matplotlib.pyplot as plt

from posydon.popsyn.synthetic_population import TransientPopulation
data_path = Path('/home/users/b/briel/scratch/high_mass_physics/data/main_figure')
```

```python
run_types = ['conservative', 'GRMHD', 'Eddington-limited']
```

### parameter setup

```python
pop_params = {
    'binary_fraction_scheme':'const',
    'binary_fraction_const': 0.7,
    
    # extending the IMF to higher and lower masses
    'primary_mass_scheme':'Kroupa2001',
    'primary_mass_min': 0.01,
    'primary_mass_max': 300,
    
    # overwriting q_min and q_max for flat mass ratio    
    'secondary_mass_scheme': 'flat_mass_ratio',
    'secondary_mass_min': 0.01,
    'secondary_mass_max': 300,
    'q_min': 0,
    'q_max': 1,
    
    # orbital parameters
    'orbital_scheme':'separation',
    'orbital_separation_scheme':'log_uniform',
    'orbital_separation_min': 5.0,
    'orbital_separation_max': 1e5,
    }


SFH_models ={
    'IllustrisTNG':{
        'delta_t' : 100, # Myr
        'SFR' : 'IllustrisTNG',
        'sigma_SFR' : None,
        'Z_min': 1e-11,
        'normalise': True,
        },
    '216f14SBBiC_FMR270_FOH_z_dM': {
        'delta_t' : 100, # Myr
        'sub_model' : '216f14SBBiC_FMR270_FOH_z_dM',
        'SFR' : 'Chruslinska+21',
        'Z_solar_scaling': 'Asplund09',
        'Z_min': 1e-11,
        'normalise': True,
    },
    '206f14SBBiC_FMR270_FOH_z_dM': {
        'delta_t' : 100, # Myr
        'sub_model' : '206f14SBBiC_FMR270_FOH_z_dM',
        'SFR' : 'Chruslinska+21',
        'Z_solar_scaling': 'Asplund09',
        'Z_min': 1e-11,
        'normalise': True,
    },
    '302f14SBBiC_FMR270_FOH_z_dM': {
        'delta_t' : 100, # Myr
        'sub_model' : '302f14SBBiC_FMR270_FOH_z_dM',
        'SFR' : 'Chruslinska+21',
        'Z_solar_scaling': 'Asplund09',
        'Z_min': 1e-11,
        'normalise': True,
    },
    'Madau_Fragos_2017': {
        'delta_t' : 100, # Myr
        'SFR' : 'Madau+Fragos17',
        'sigma' : 0.5,
        'Z_min': 1e-11,
        'normalise': True,
    },
    'Neijssel_2019':{
        'delta_t' : 100, # Myr
        'SFR' : 'Neijssel+19',
        'sigma' : 0.39,
        'Z_min': 1e-11,
        'normalise': True,
    },
}
```

```python
for run in run_types:
    print(run)
    
    file = data_path / f"{run}.h5" 
    transient_pop = TransientPopulation(str(file), 'BBH')
    
    # The same for the main_figure populations
    sim_params = transient_pop.ini_params.copy()
    sim_params['orbital_separation_scheme'] = 'log_uniform'
    sim_params['orbital_separation_min'] = 5.0
    sim_params['orbital_separation_max'] = 1e5
    
    rates = transient_pop.calculate_model_weights(model_weights_identifier='extended_IMF',
                                                  population_parameters=pop_params,
                                                  simulation_parameters=sim_params)
    
    for model in SFH_models.keys():
        print(model)
        rates = transient_pop.calculate_cosmic_weights(model,
                                                       model_weights='extended_IMF',
                                                       MODEL_in=SFH_models[model])
        rates.calculate_intrinsic_rate_density(channels=True)
```
