```python
import os
data_dir = '/home/users/b/briel/scratch/high_mass_physics/data/GWTC_posteriors'

out_paths = []
for event_name in event_names:
    print(event_name)
    files_in_dir = os.listdir(data_dir)

    # check for any filename that contains the event name substring
    matches = [fn for fn in files_in_dir if event_name in fn]
    if matches:
        file_path = os.path.join(data_dir, matches[0])
        print("Found existing file:", file_path)
    else:
        
        if int(event_name[2:4]) < 18:
            event_name = event_name.split('_')[0]
            try:
                file_path = fetch_open_samples(event_name,
                                           outdir=data_dir,
                                           verbose=False)
            except:
                print("Could not find", event_name)
        
        elif int(event_name[2:4]) < 21:
            catalog = 'GWTC-2'
            try:
                file_path = fetch_open_samples(event_name,
                                           unpack=True,
                                           read_file=False,
                                           delete_on_exit=False,
                                           outdir=data_dir,
                                           catalog=catalog,
                                           verbose=False)
            except:
                event_name = event_name.split('_')[0]
                file_path = fetch_open_samples(event_name,
                                             unpack=True,
                                             read_file=False,
                                             delete_on_exit=False,
                                             outdir=data_dir,
                                             catalog=catalog,
                                             verbose=False)
                                               

        else:
            file_path = fetch_open_samples(event_name,
                                            unpack=False,
                                            read_file=False,
                                            delete_on_exit=False,
                                            outdir=data_dir, verbose=False)
    out_paths.append(file_path)
```
